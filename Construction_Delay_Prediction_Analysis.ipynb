```python
import pandas as pd

df = pd.read_csv('Mini_Project_Construction_Delay_Dataset.csv')

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Preprocessing Step 2: Check Missing Values

df.isnull().sum()
```

```python
# Preprocessing Step 3: Check Duplicate Records
df.duplicated().sum()
```

```python
# Preprocessing Step 4: Identify Categorical Features

categorical_columns = df.select_dtypes(include=['object', 'str']).columns

categorical_columns
```

```python
# Preprocessing Step 5: Check Target Values

df['Project_Delayed'].value_counts()
```

```python
# Preprocessing Step 6: Encode Target Variable

y = df['Project_Delayed'].map({
    'No': 0,
    'Yes': 1
})

y.value_counts()
```

```python
# Preprocessing Step 7: Prepare Input Features

X = df.drop(columns=['Project_ID', 'Project_Delayed'])

X.head()
```

```python
# Preprocessing Step 8: Encode Project Type

X = pd.get_dummies(X, columns=['Project_Type'], drop_first=True)

X.head()
```

```python
# Preprocessing Step 9: Check Feature Data Types

X.dtypes
```

```python
# Preprocessing Step 10: Convert Boolean to Integer

X['Project_Type_Residential'] = X['Project_Type_Residential'].astype(int)

X.dtypes
```

```python
# Preprocessing Step 11: Final Feature Check

print("Feature shape:", X.shape)
print("Missing values:", X.isnull().sum().sum())
print("Target shape:", y.shape)
```

```python
# Preprocessing Step 12: Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Preprocessing Step 12: Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Preprocessing Step 13: Verify Target Distribution

print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Preprocessing Step 14: Check Feature Ranges

X_train.describe().T[['min', 'max']]
```

```python
# Preprocessing Step 14: Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

```python
# Model 1: Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

logistic_model = LogisticRegression(random_state=42)

logistic_model.fit(X_train_scaled, y_train)

y_pred_logistic = logistic_model.predict(X_test_scaled)

print("Logistic Regression Accuracy:",
      accuracy_score(y_test, y_pred_logistic))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic))
```

```python
# Model 2: Decision Tree

from sklearn.tree import DecisionTreeClassifier

decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

decision_tree_model.fit(X_train, y_train)

y_pred_tree = decision_tree_model.predict(X_test)

print("Decision Tree Accuracy:",
      accuracy_score(y_test, y_pred_tree))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree))
```

```python
# Model 3: Random Forest

from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

random_forest_model.fit(X_train, y_train)

y_pred_rf = random_forest_model.predict(X_test)

print("Random Forest Accuracy:",
      accuracy_score(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# Model 4: Support Vector Machine

from sklearn.svm import SVC

svm_model = SVC(
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)

y_pred_svm = svm_model.predict(X_test_scaled)

print("SVM Accuracy:",
      accuracy_score(y_test, y_pred_svm))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_svm))
```

```python
# Step 15: Confusion Matrix - Logistic Regression

cm_logistic = confusion_matrix(y_test, y_pred_logistic)

print("Confusion Matrix:")
print(cm_logistic)
```

```python
# Step 16: ROC-AUC and Delay Probability

from sklearn.metrics import roc_auc_score

y_prob_logistic = logistic_model.predict_proba(X_test_scaled)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob_logistic)

print("ROC-AUC Score:", roc_auc)

print("\nFirst 10 Delay Probabilities:")
print(y_prob_logistic[:10])
```

```python
# Step 17: Convert Delay Probability into Risk Level

def get_risk_level(probability):
    if probability < 0.30:
        return "Low"
    elif probability < 0.60:
        return "Medium"
    elif probability < 0.80:
        return "High"
    else:
        return "Very High"


risk_levels = [get_risk_level(p) for p in y_prob_logistic]

print("First 10 Risk Levels:")
print(risk_levels[:10])
```

```python
# Step 18: Risk Level Distribution

from collections import Counter

risk_distribution = Counter(risk_levels)

print("Risk Level Distribution:")
for risk, count in risk_distribution.items():
    print(f"{risk}: {count}")
```

```python
# Step 19: Model Performance Comparison

model_comparison = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest',
        'SVM'
    ],
    'Accuracy': [
        accuracy_score(y_test, y_pred_logistic),
        accuracy_score(y_test, y_pred_tree),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_svm)
    ]
})

model_comparison['Accuracy_Percent'] = (
    model_comparison['Accuracy'] * 100
).round(2)

print(model_comparison)
```

```python
# Step 20: Feature Importance using Logistic Regression Coefficients

feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Coefficient': logistic_model.coef_[0]
})

feature_importance['Absolute_Coefficient'] = (
    feature_importance['Coefficient'].abs()
)

feature_importance = feature_importance.sort_values(
    by='Absolute_Coefficient',
    ascending=False
)

print("Feature Importance:")
print(feature_importance)
```

```python
# Step 21: Cross-Validation of Logistic Regression

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    logistic_model,
    X_train_scaled,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("Cross-Validation Accuracy Scores:")
print(cv_scores)

print("\nMean Cross-Validation Accuracy:",
      cv_scores.mean())

print("Standard Deviation:",
      cv_scores.std())
```

```python
def predict_project_delay(project_data):
    
    # Convert input into DataFrame
    input_df = pd.DataFrame([project_data])
    
    # One-hot encode Project_Type
    input_df = pd.get_dummies(
        input_df,
        columns=['Project_Type'],
        drop_first=True
    )
    
    # Ensure all training features are present
    input_df = input_df.reindex(columns=X.columns, fill_value=0)
    
    # Scale the input
    input_scaled = scaler.transform(input_df)
    
    # Predict probability
    delay_probability = logistic_model.predict_proba(input_scaled)[0][1]
    
    # Predict delay
    prediction = logistic_model.predict(input_scaled)[0]
    
    # Determine risk level
    risk_level = get_risk_level(delay_probability)
    
    return {
        "Delay Probability": round(delay_probability * 100, 2),
        "Risk Level": risk_level,
        "Predicted Delay": "Yes" if prediction == 1 else "No"
    }
```

```python
sample_project = {
    'Project_Type': 'Residential',
    'Project_Area_sqft': 2500,
    'Number_of_Floors': 2,
    'Planned_Duration_Months': 12,
    'Planned_Manpower': 30,
    'Average_Manpower': 25,
    'Material_Availability_Percent': 90,
    'Equipment_Availability_Percent': 92,
    'Material_Delay_Days': 5,
    'Procurement_Delay_Days': 6,
    'Weather_Delay_Days': 2,
    'Change_Order_Count': 3,
    'Planned_Progress_Percent': 70,
    'Actual_Progress_Percent': 55,
    'Progress_Gap_Percent': 15
}

result = predict_project_delay(sample_project)

print("Prediction Result:")
print(result)
```

```python
import streamlit as st
import joblib

# --------------------------------------------------
# Page Configuration
# --------------------------------------------------

st.set_page_config(
    page_title="Construction Delay Prediction",
    page_icon="🏗️",
    layout="wide"
)

# --------------------------------------------------
# Load Model and Scaler
# --------------------------------------------------

model = joblib.load("logistic_delay_model.pkl")
scaler = joblib.load("delay_scaler.pkl")

# --------------------------------------------------
# Reset Function
# --------------------------------------------------

def reset_inputs():
    st.session_state["project_type"] = "Residential"
    st.session_state["project_area"] = 2500
    st.session_state["number_of_floors"] = 2
    st.session_state["planned_duration"] = 12
    st.session_state["planned_manpower"] = 30
    st.session_state["average_manpower"] = 25
    st.session_state["material_availability"] = 90.0
    st.session_state["equipment_availability"] = 92.0
    st.session_state["material_delay"] = 4
    st.session_state["procurement_delay"] = 4
    st.session_state["weather_delay"] = 2
    st.session_state["change_orders"] = 2
    st.session_state["planned_progress"] = 60.0
    st.session_state["actual_progress"] = 50.0


# --------------------------------------------------
# Title
# --------------------------------------------------

st.title("🏗️ Construction Project Delay Prediction")

st.write(
    "AI-Based Construction Project Delay Prediction and Risk Analysis System"
)

# --------------------------------------------------
# About the Project
# --------------------------------------------------

with st.expander("ℹ️ About This Project"):
    st.write("""
    This application uses Machine Learning to predict whether a
    construction project is likely to experience a delay.

    The system analyzes project details such as project size,
    manpower, material and equipment availability, delays,
    change orders, and progress information.

    A Logistic Regression model is used to generate the delay
    prediction and probability.

    Based on the predicted probability, the project is categorized
    into Low, Medium, or High Risk.

    Technologies Used:
    • Python
    • Pandas
    • NumPy
    • Scikit-learn
    • Streamlit
    • Joblib
    """)

# --------------------------------------------------
# Project Details
# --------------------------------------------------

st.header("📋 Project Details")

# --------------------------------------------------
# Basic Project Information
# --------------------------------------------------

st.subheader("🏗️ Basic Project Information")

col1, col2 = st.columns(2)

with col1:

    project_type = st.selectbox(
        "Project Type",
        ["Residential", "Commercial"],
        key="project_type"
    )

    project_area = st.number_input(
        "Project Area (sqft)",
        min_value=500,
        max_value=100000,
        value=2500,
        key="project_area"
    )

    number_of_floors = st.number_input(
        "Number of Floors",
        min_value=1,
        max_value=50,
        value=2,
        key="number_of_floors"
    )

with col2:

    planned_duration = st.number_input(
        "Planned Duration (Months)",
        min_value=1,
        max_value=120,
        value=12,
        key="planned_duration"
    )

    planned_manpower = st.number_input(
        "Planned Manpower",
        min_value=1,
        max_value=200,
        value=30,
        key="planned_manpower"
    )

    average_manpower = st.number_input(
        "Average Manpower",
        min_value=1,
        max_value=200,
        value=25,
        key="average_manpower"
    )

# --------------------------------------------------
# Resource and Delay Information
# --------------------------------------------------

st.subheader("📦 Resource & Delay Information")

col3, col4 = st.columns(2)

with col3:

    material_availability = st.number_input(
        "Material Availability (%)",
        min_value=0.0,
        max_value=100.0,
        value=90.0,
        key="material_availability"
    )

    equipment_availability = st.number_input(
        "Equipment Availability (%)",
        min_value=0.0,
        max_value=100.0,
        value=92.0,
        key="equipment_availability"
    )

    material_delay = st.number_input(
        "Material Delay (Days)",
        min_value=0,
        max_value=100,
        value=4,
        key="material_delay"
    )

with col4:

    procurement_delay = st.number_input(
        "Procurement Delay (Days)",
        min_value=0,
        max_value=100,
        value=4,
        key="procurement_delay"
    )

    weather_delay = st.number_input(
        "Weather Delay (Days)",
        min_value=0,
        max_value=100,
        value=2,
        key="weather_delay"
    )

    change_orders = st.number_input(
        "Change Order Count",
        min_value=0,
        max_value=50,
        value=2,
        key="change_orders"
    )

# --------------------------------------------------
# Progress Details
# --------------------------------------------------

st.subheader("📊 Progress Details")

col5, col6 = st.columns(2)

with col5:

    planned_progress = st.number_input(
        "Planned Progress (%)",
        min_value=0.0,
        max_value=100.0,
        value=60.0,
        key="planned_progress"
    )

with col6:

    actual_progress = st.number_input(
        "Actual Progress (%)",
        min_value=0.0,
        max_value=100.0,
        value=50.0,
        key="actual_progress"
    )

# --------------------------------------------------
# Progress Gap
# --------------------------------------------------

progress_gap = planned_progress - actual_progress

if actual_progress > planned_progress:

    st.error(
        "⚠️ Actual Progress cannot be greater than Planned Progress."
    )

else:

    st.info(
        f"📊 Progress Gap: **{progress_gap:.2f}%**"
    )

# --------------------------------------------------
# Buttons
# --------------------------------------------------

st.success("Project details entered successfully.")

button_col1, button_col2 = st.columns([1, 5])

with button_col1:

    if st.button("🔮 Predict Project Delay"):

        if actual_progress > planned_progress:

            st.error(
                "Prediction cannot be performed because Actual Progress "
                "cannot be greater than Planned Progress."
            )

            st.stop()

        # --------------------------------------------------
        # Prepare Input Data
        # --------------------------------------------------

        project_type_residential = (
            1 if project_type == "Residential" else 0
        )

        input_data = [[
            project_area,
            number_of_floors,
            planned_duration,
            planned_manpower,
            average_manpower,
            material_availability,
            equipment_availability,
            material_delay,
            procurement_delay,
            weather_delay,
            change_orders,
            planned_progress,
            actual_progress,
            progress_gap,
            project_type_residential
        ]]

        # --------------------------------------------------
        # Scale Input
        # --------------------------------------------------

        scaled_data = scaler.transform(input_data)

        # --------------------------------------------------
        # Prediction
        # --------------------------------------------------

        prediction = model.predict(scaled_data)[0]

        probability = (
            model.predict_proba(scaled_data)[0][1] * 100
        )

        # --------------------------------------------------
        # Prediction Text
        # --------------------------------------------------

        if prediction == 1:

            prediction_text = "DELAYED"

        else:

            prediction_text = "ON TIME"

        # --------------------------------------------------
        # Risk Level
        # --------------------------------------------------

        if probability < 40:

            risk_level = "Low Risk 🟢"

        elif probability <= 70:

            risk_level = "Medium Risk 🟡"

        else:

            risk_level = "High Risk 🔴"

        # --------------------------------------------------
        # Prediction Result
        # --------------------------------------------------

        st.divider()

        st.header("🎯 Prediction Result")

        result_col1, result_col2, result_col3 = st.columns(3)

        with result_col1:

            st.metric(
                "Project Status",
                prediction_text
            )

        with result_col2:

            st.metric(
                "Delay Probability",
                f"{probability:.2f}%"
            )

        with result_col3:

            st.metric(
                "⚠️ Risk Level",
                risk_level
            )

        # --------------------------------------------------
        # Risk Level Explanation
        # --------------------------------------------------

        if probability < 40:

            st.success(
                "🟢 Low Risk: The project has a relatively low "
                "probability of delay."
            )

        elif probability <= 70:

            st.warning(
                "🟡 Medium Risk: The project has a moderate "
                "probability of delay. Project factors should "
                "be monitored carefully."
            )

        else:

            st.error(
                "🔴 High Risk: The project has a high probability "
                "of delay. Important delay factors should be reviewed."
            )

        # --------------------------------------------------
        # Probability Visualization
        # --------------------------------------------------

        st.subheader("📈 Delay Probability Visualization")

        st.progress(int(probability))

        st.write(
            f"Delay Probability: **{probability:.2f}%**"
        )

        # --------------------------------------------------
        # Project Summary
        # --------------------------------------------------

        st.subheader("📋 Project Summary")

        summary_col1, summary_col2, summary_col3 = st.columns(3)

        with summary_col1:

            st.write(f"**Project Type:** {project_type}")
            st.write(f"**Project Area:** {project_area:.0f} sqft")
            st.write(f"**Number of Floors:** {number_of_floors}")
            st.write(f"**Planned Duration:** {planned_duration} months")
            st.write(f"**Planned Manpower:** {planned_manpower}")

        with summary_col2:

            st.write(f"**Average Manpower:** {average_manpower}")
            st.write(
                f"**Material Availability:** "
                f"{material_availability:.1f}%"
            )
            st.write(
                f"**Equipment Availability:** "
                f"{equipment_availability:.1f}%"
            )
            st.write(f"**Material Delay:** {material_delay} days")
            st.write(
                f"**Procurement Delay:** "
                f"{procurement_delay} days"
            )

        with summary_col3:

            st.write(f"**Weather Delay:** {weather_delay} days")
            st.write(f"**Change Orders:** {change_orders}")
            st.write(
                f"**Planned Progress:** "
                f"{planned_progress:.1f}%"
            )
            st.write(
                f"**Actual Progress:** "
                f"{actual_progress:.1f}%"
            )
            st.write(
                f"**Progress Gap:** "
                f"{progress_gap:.1f}%"
            )

        # --------------------------------------------------
        # Prediction Interpretation
        # --------------------------------------------------

        st.subheader("📝 Prediction Interpretation")

        if prediction == 1:

            st.warning(
                "The project has a higher likelihood of experiencing "
                "delay. Review manpower, material availability, "
                "procurement, weather, and progress-related factors."
            )

        else:

            st.info(
                "The project has a lower likelihood of experiencing "
                "delay based on the entered project details."
            )

        # --------------------------------------------------
        # Model Information
        # --------------------------------------------------

        with st.expander("🤖 Model Information"):

            st.write(
                "**Machine Learning Algorithm:** Logistic Regression"
            )

            st.write(
                "**Purpose:** Predict construction project delay"
            )

            st.write(
                "**Output:** Delay status and delay probability"
            )

            st.write(
                "**Risk Categories:** Low Risk, Medium Risk, High Risk"
            )

            st.write(
                "**Input Features:** 15 project-related features"
            )


with button_col2:

    if st.button("🔄 Clear / Reset", on_click=reset_inputs):

        st.rerun()

# --------------------------------------------------
# Footer
# --------------------------------------------------

st.divider()

st.caption(
    "AI-Based Construction Project Delay Prediction and Risk Analysis System"
)

st.caption(
    "Developed as an MCA Mini Project using Machine Learning and Streamlit"
)
```
